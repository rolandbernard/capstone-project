First we use the SALSA dataset as a demo, since it is a very difficult situation for the system.

```python
data = dataset.SalsaDataset(path="../data/salsa")

source = data.get_source()
source.scale(100.0)
source.to(util.DEVICE);
```

```python
track = tracker.build_tuned(False, False, False, path="..")
track.mo_threshold, track.mn_threshold = -2, -5
cams, frames, fps = track.evaluate(source, limit=4500)
util.save_tracks("../stats/demo/salsa-baseline.json", cams,
                 frames, fps, (1000, 300, 0), (0, 0, 1))
```

```python
track = tracker.build_tuned(False, False, path="..")
track.mo_threshold, track.mn_threshold = -2, -5
cams, frames, fps = track.evaluate(source, limit=4500)
util.save_tracks("../stats/demo/salsa-constrained.json", cams,
                 frames, fps, (1000, 300, 0), (0, 0, 1))
```

```python
with open(f"../stats/demo/salsa-baseline.json", "rb") as f:
    f_data = f.read()
with open("../stats/demo/salsa.json", "wb") as f:
    f.write(f_data)

source = data.get_source()
source.scale(100.0)
demo.augment_tracks(
    "../stats/demo/salsa.json", source.cameras, source.streams,
    True, False, (1000, 300, 0), (0, 0, 1))
```

Next we use some panoptic dataset test scenes for the demo. First argument the ground truth recordings using the video URLs and a point cloud of the room.

```python
data = dataset.CmuPanopticDataset(path="../data/panoptic")
```

```python
for scene in data.test_scenes:
    source = data.get_source(scene, num_vga_cams=16)
    demo.augment_tracks(
        f"../stats/gt/{scene}.json", source.cameras, source.streams,
        True, False, (0, 0, 0), (0, -1, 0))
```

```python
for scene, name in [
    ("160906_pizza1", "pizza"), ("161029_sports1", "sports"),
    ("160224_haggling1", "haggling"), ("161029_build1", "build"),
]:
    with open(f"../stats/gt/{scene}.json", "rb") as f:
        f_data = f.read()
    with open(f"../stats/gt/{name}.json", "wb") as f:
        f.write(f_data)
    source = data.get_source(scene, num_vga_cams=16)
    demo.augment_tracks(
        f"../stats/gt/{name}.json", source.cameras, source.streams,
        True, False, (0, 0, 0), (0, -1, 0))
```

```python
scene = "160224_haggling1"

with open(f"../stats/gt/{scene}.json", "rb") as f:
    f_data = f.read()
with open("../stats/gt/full.json", "wb") as f:
    f.write(f_data)
source = data.get_source(scene, num_vga_cams=8)
demo.augment_tracks(
    "../stats/gt/full.json", source.cameras, source.streams,
    True, False, (0, 0, 0), (0, -1, 0))
```
